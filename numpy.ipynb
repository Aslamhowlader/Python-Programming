```python
'''. Array Creation

NumPy arrays can be created in many ways:

np.array() → convert list to array
np.zeros(shape) → all 0s
np.ones(shape) → all 1s
np.empty(shape) → uninitialized values
np.arange(start, stop, step) → range-based array
np.linspace(start, stop, num) → equal intervals
np.random.random(n) → random floats (0–1)
np.random.randint(low, high, size) → random integers
===============================================================
2. Array Shapes & Properties
a.shape → dimensions of array
a.ndim → number of dimensions
a.size → total elements
a.dtype → data type
a.reshape() → change shape
==============================================
3. Indexing & Slicing
1D:
a[start:stop:step]
Negative indexing: a[-1]
Reverse: a[::-1]
2D:
a[row, col]
a[row, :] → full row
a[:, col] → full column
Submatrix slicing: a[r1:r2, c1:c2]
Boolean indexing:
a[a > 10]
a[(a > 10) & (a % 2 == 0)]
========================================
4. Array Operations
Scalar operations:
a + 5, a - 5, a * 2, a / 2
Comparison:
a > 5, a == 10, etc. (returns boolean array)
=========================================
5. Mathematical Functions
np.sum()
np.prod()
np.mean()
np.median()
np.std()
np.var()
np.min(), np.max()
Axis concept:
axis=0 → column-wise
axis=1 → row-wise
==============================================
6. Trigonometric Functions
np.sin()
np.cos()
np.tan()

(operate element-wise on arrays)
============================================

7. Exponential & Log Functions
np.exp() → e^x
np.log() → natural log
np.log10() → base-10 log
np.sqrt() → square root

⚠️ log only works for positive values
=====================================================

8. Linear Algebra
np.dot(A, B) → matrix multiplication
(Requires compatible shapes)
==========================================
9. Set Operations
np.union1d() → union
np.intersect1d() → intersection
np.setdiff1d() → difference
np.setxor1d() → symmetric difference
===========================================
10. Array Manipulation
np.concatenate() → join arrays
np.vstack() → vertical stack
np.hstack() → horizontal stack
np.split() → split array
np.expand_dims() → add dimension
np.flip() → reverse array
===============================================
11. Utility Functions
np.where() → conditional selection
np.unique() → unique values
np.isin() → membership check
np.percentile() → percentile value
np.clip() → limit values
np.put() → modify elements by index
=====================================================
12. Sorting & Searching
np.sort() → sort values
np.argsort() → sorted indices
np.argmax() → index of max
np.argmin() → index of min
======================================
13. Plotting (Matplotlib with NumPy)

You plotted:

y = x
y = x²
y = sin(x)
y = x log(x)

Key idea:

np.linspace() creates smooth x-axis values
matplotlib visualizes NumPy computations

========================================

14. Key Concepts You Practiced
Vectorized operations (no loops)
Broadcasting
Axis-based computation
Boolean filtering
Matrix operations
Data reshaping'''
```
